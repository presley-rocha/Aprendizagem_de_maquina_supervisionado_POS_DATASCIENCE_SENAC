```python
import pandas as pd
```

```python
nomeArquivo_2023=('C:/Users/Bruno/Desktop/Pos_SENAC_DF/Aprendizagem_de_maquina_supervisionado/DataSets/acidentes2023.csv')
nomeArquivo_2024=('C:/Users/Bruno/Desktop/Pos_SENAC_DF/Aprendizagem_de_maquina_supervisionado/DataSets/acidentes2024.csv')
nomeArquivo_2025=('C:/Users/Bruno/Desktop/Pos_SENAC_DF/Aprendizagem_de_maquina_supervisionado/DataSets/acidentes2025.csv')
nomeArquivo_2026=('C:/Users/Bruno/Desktop/Pos_SENAC_DF/Aprendizagem_de_maquina_supervisionado/DataSets/acidentes2026.csv')
```

```python
df_2023=pd.read_csv(nomeArquivo_2023, sep=";", encoding="latin1")
df_2024=pd.read_csv(nomeArquivo_2024, sep=";", encoding="latin1")
df_2025=pd.read_csv(nomeArquivo_2025, sep=";", encoding="latin1")
df_2026=pd.read_csv(nomeArquivo_2026, sep=";", encoding="latin1")
```

Vamos comparar todas as variáveis dos 4 diferentes Datasets para de cada ano para avaliar se todos estão com as mesmas disposições e dtypes das variáveis.

```python
dfs = {'df_2023': df_2023, 'df_2024': df_2024, 'df_2025': df_2025, 'df_2026': df_2026}
df_dtypes = pd.DataFrame({name: df.dtypes for name, df in dfs.items()})
df_dtypes = df_dtypes.fillna('N/A')
print(df_dtypes)
```

Vamos agora unir todos os Dataset de 'forma empilhada' para trabalharmos somente com um único Dataset, vamos usar para isso a função pd.concat()

```python
lista_dataset = [df_2023, df_2024, df_2025, df_2026]
Dataset_unificado = pd.concat(lista_dataset, ignore_index=True)
print(Dataset_unificado.info())
```

```python
Dataset_unificado = Dataset_unificado.drop_duplicates()
print(Dataset_unificado.shape)
```

Vamos salvar no diretório o Dataset_unificado.csv com todos os anos já agrupados, qua será utilizado agora no nosso trabalho.

```python
Dataset_unificado.to_csv('C:/Users/Bruno/Desktop/Pos_SENAC_DF/Aprendizagem_de_maquina_supervisionado/DataSets/Dataset_unificado.csv', index=False, sep=';', encoding='latin1')
```
