## Vamos ler diretamente para o Pandas (Sem descompactar no disco)
Esta é a melhor prática em ciência de dados para evitar redundância de arquivos e economizar espaço de armazenamento:

```python
import pandas as pd
import io
import zipfile
```

```python
caminho_zip = 'C:/Users/Bruno/Desktop/Pos_SENAC_DF/Aprendizagem_de_maquina_supervisionado/Trabalho_Final/1_DataSet/Dataset_unificado.zip'

with zipfile.ZipFile(caminho_zip, 'r') as zip_ref:
  nome_csv = [f for f in zip_ref.namelist() if f.endswith('.csv')][0]

  with zip_ref.open(nome_csv) as arquivo_csv:
    df = pd.read_csv(arquivo_csv, sep=None, engine='python', encoding='latin1')

print(df.info())
```
